```python
class Link:

  def __init__(self):
    self.a = 1
    self.b = 2
```

```python
l = Link()

l.a
```

```python
l.b
```

```python
class Chain(Link):

  def sum(self):
    return self.a + self.b
```

```python
# Chain クラスをインスタンス化
c = Chain()

c.a
```

```python
c.b
```

```python
# sum メソッドを実行
c.sum()
```

```python
class Chain(Link):
  def __init__(self):
    self.c = 5 # self.c を新たに追加

  def sum(self):
    return self.a + self.b + self.c

# インスタンス化
C = Chain()
```

```python
class Chain(Link):

  def __init__(self):
    # 親クラスの __init__()　メソッドを呼び出す
    super().__init__()

    # self.c を新たに追加
    self.c = 5

  def sum(self):
    return self.a + self.b + self.c

# インスタンス化
c = Chain()
```

```python
c.sum()
```
